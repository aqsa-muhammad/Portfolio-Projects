```python
#import Libraries
from bs4 import BeautifulSoup
import requests
```

```python
# Connect to website
url = 'https://en.wikipedia.org/wiki/List_of_largest_companies_in_the_United_States_by_revenue'
headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'}
```

```python
page = requests.get(url,headers=headers)
soup=BeautifulSoup(page.text,'html.parser')
```

```python
#print(soup.prettify())
```

```python
#Find the 1st table from all 3 tables via th tag of TABLE HEADER
table = soup.find_all('table')[0]
```

```python
#world_titles = soup.find_all('th')
# world_titles
# now this will get all data of th from page we want from table only which is table  number 1
world_titles = table.find_all('th')
world_titles
```

```python
#Do list comprehension
world_table_titles= [title.text for title in world_titles]
print(world_table_titles)
```

```python
import pandas as pd
df = pd.DataFrame(columns = world_table_titles)
df
```

```python
column_data = table.find_all('tr')
#column_data
```

```python
# for row in column_data: 
    # row_data = row.find_all('td')
    # individual_row_data = [data.text for data in row_data]
    # print(individual_row_data)
    
```

```python
#loops through each table row (skipping the header), extracts the text from every cell (<td>), 
#and appends that row as a new entry to the bottom of your Pandas DataFrame.
for row in column_data[1:]: 
    row_data = row.find_all('td')
    individual_row_data = [data.text for data in row_data]
        
    length = len(df)
    df.loc[length] = individual_row_data
    
```
